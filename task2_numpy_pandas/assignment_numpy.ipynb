**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
arr = np.random.randint(0, 10, size=10)

new_arr = np.where((arr > 3) & (arr < 8), -arr, arr)
print(arr)
print(new_arr)

```

**2.** Заменить максимальный элемент случайного массива на 0

```python
arr = np.random.randint(0, 10, size=10)

new_arr = np.where(arr == arr.max(), 0, arr)

print(arr)
print(new_arr)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
arr = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
])

grids = np.meshgrid(*arr, indexing='ij')
stacked = np.stack(grids, axis=-1)
new_arr = stacked.reshape(-1, arr.shape[0])

print(arr)
print(new_arr)
```

**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
A = np.array([
    [1, 3, 2],
    [2, 3, 4],
    [1, 2, 4],
    [2, 3, 5],
    [1, 3, 4],
    [0, 1, 2],
    [2, 3, 6],
    [1, 2, 5]
])

B = np.array([
    [1, 2],
    [2, 3]
])

eq = A[:, :, None, None] == B[None, None, :, :]

contains_each_B_row = eq.any(axis=1).all(axis=2)

mask = contains_each_B_row.all(axis=1)

result = A[mask]
print(mask)
print(result)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
arr = np.array([
    [1, 3, 2],
    [2, 3, 4],
    [1, 2, 4],
    [2, 3, 5],
    [1, 3, 4],
    ['a', 'a', 'a'],
    [9, 9, 9],
    [0, 0, 0],
    [1, 1, 1],
    [2, 2, 2]
])

same_as_first = (arr == arr[:, [0]]).all(axis=1)
mask = ~same_as_first
result = arr[mask]

print(arr)
print()
print(result)
```

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
X = [
    [1, 0, 1],
    [2, 0, 2],
    [3, 0, 3],
    [4, 4, 4]
]

## Python way
mult1 = 1
for i in range(min(len(X), len(X[0]))):
    if X[i][i] != 0:
        mult1 *= X[i][i]
print(mult1)

## Numpy way
d = np.diag(X)
mult2 = np.prod(d[d != 0])

print(mult2)
```

* __Задача 2__: Даны два вектора x и y. Проверить, задают ли они одно и то же мультимножество.  
  Например, для x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) ответ True.

```python
x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

## Python way
flag = False
if len(x) == len(y) and sorted(x) == sorted(y):
    flag = True
print(flag)

## Numpy way
flag2 = False
x = np.asarray(x)
y = np.asarray(y)
if x.size != y.size:
    flag2 = False
if x.size == y.size and np.array_equal(np.sort(x), np.sort(y)):
    flag2 = True
print(flag2)
    
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль. 
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]

## Python way
after_zero = []
for i in range(1, len(x)):
    if x[i - 1] == 0:
        after_zero.append(x[i])
print(max(after_zero))

## Numpy way
x = np.asarray(x)
mask = (x[:-1] == 0)
result = x[1:][mask].max()
print(result)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = [2, 2, 2, 3, 3, 3, 5]

## Python way
values = []
summ = []
current = x[0]
cnt = 1

for i in range(1, len(x)):
    if x[i] == current:
        cnt += 1
    else:
        values.append(current)
        summ.append(cnt)
        current = x[i]
        cnt = 1

values.append(current)
summ.append(cnt)

print(values) 
print(summ)

## Numpy way
x = np.asarray(x)
indexes = np.where(np.diff(x) != 0)[0] + 1
values2 = x[np.concatenate(([0], indexes))]
summ2 = np.diff(np.concatenate(([0], indexes, [x.size])))

print(values2) 
print(summ2)
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import math
from scipy.spatial.distance import cdist
from timeit import timeit

x = [(1, 2), (3, 4)]
y = [(5, 6), (7, 8)]

## Python way
result1 = []

for a in x:
    row = []
    for b in y:
        sq = 0
        for i in range(len(b)):
            diff = a[i] -b[i]
            sq += diff * diff
        row.append(math.sqrt(sq))
    result1.append(row)
print("Python result:")
print(result1)

## Numpy way

xx = np.array(x)
yy = np.array(y)

diff = xx[:, None, :] - yy[None, :, :]
sq2 = diff ** 2
result2 = np.sqrt(sq2.sum(axis=2))

print("Numpy result:")
print(result2)


## Speed test

n, m, d = 1000, 1000, 32
X = np.random.randn(n, d)
Y = np.random.randn(m, d)

def numpy_dist():
    diff = X[:, None, :] - Y[None, :, :]
    return np.sqrt((diff * diff).sum(axis=2))

def scipy_dist():
    return cdist(X, Y, metric="euclidean")


t_np = timeit(numpy_dist, number=5)
t_sp = timeit(scipy_dist, number=5)
print()
print("NumPy spped:", t_np, "sec")
print("SciPy speed:", t_sp, "sec")
```
