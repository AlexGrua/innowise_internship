**4-5. Рассчитайте среднее значение и среднеквадратичное отклонение возраста тех, кто получает более 50K в год (признак *salary*) и тех, кто получает менее 50K в год**

```python
df = data
avg_age_over_50K = df.groupby('salary')['age'].mean().loc['>50K']
std_over_50K = df.groupby('salary')['age'].std().loc['>50K']
avg_age_under_50K = df.groupby('salary')['age'].mean().loc['<=50K']
std_under_50K = df.groupby('salary')['age'].std().loc['<=50K']

print("Average age for salary over 50K:", avg_age_over_50K)
print("Average age for salary under 50K:", avg_age_under_50K)
print("Standard deviation on age for salary over 50K:", std_over_50K)
print("Standard deviation on age for salary under 50K:", std_under_50K)

```

**6. Правда ли, что люди, которые получают больше 50k, имеют минимум высшее образование? (признак *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

```python
df = data
higher_edu = [
    'Bachelors', 'Prof-school', 'Assoc-acdm',
    'Assoc-voc', 'Masters', 'Doctorate'
]
total_over_50K = df['salary'].value_counts().loc['>50K']
high_edu_for_over_50K = df[df['salary']=='>50K'].loc[df['education'].isin(higher_edu), 'education'].value_counts()
total_high_edu_for_over_50K = df[df['salary']=='>50K'].loc[df['education'].isin(higher_edu), 'education'].count()
perc_high_edu_over_50K = high_edu_for_over_50K/total_over_50K*100

print("High education percentage for salary over 50K:", total_high_edu_for_over_50K/total_over_50K*100)
print()

print("Percentage per degree for salary over 50K", perc_high_edu_over_50K)

print("Большинство людей (57%) с зп свыше 50К имеют минимум высшее образование")
```

**7. Выведите статистику возраста для каждой расы (признак *race*) и каждого пола. Используйте *groupby* и *describe*. Найдите таким образом максимальный возраст мужчин расы *Asian-Pac-Islander*.**

```python
df = data

stats = df.groupby(['race', 'sex'])['age'].describe()

max_age = stats.loc[('Asian-Pac-Islander', 'Male'), 'max']

print(stats)
print()
print("Максимальный возраст мужчин Asian-Pac-Islander:", max_age)
```

**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
df = data
married = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']

total_married = df[(df['sex'] == 'Male') &  (df['marital-status'].isin(married))]['salary'].count()
total_married_over_50K = df[(df['salary'] == '>50K') & (df['sex'] == 'Male') &  (df['marital-status'].isin(married))]['salary'].count()
married_over_50K_perc = total_married_over_50K/total_married*100

total_not_married = df[(df['sex'] == 'Male') &  (~df['marital-status'].isin(married))]['salary'].count()
total_not_married_over_50K = df[(df['salary'] == '>50K') & (df['sex'] == 'Male') &  (~df['marital-status'].isin(married))]['salary'].count()
not_married_over_50K_perc = total_not_married_over_50K/total_not_married*100


print("Процент зарабатывающих более 50К среди женатых:", married_over_50K_perc)
print("Процент зарабатывающих более 50К среди холостых:", not_married_over_50K_perc)
```

**9. Какое максимальное число часов человек работает в неделю (признак *hours-per-week*)? Сколько людей работают такое количество часов и каков среди них процент зарабатывающих много?**

```python
df = data
max_hours = df['hours-per-week'].max()
people_with_max_hours = (df['hours-per-week']== max_hours).sum()

people_with_max_hours_over_50K = ((df['hours-per-week']== max_hours) & (df['salary'] == '>50K') ).sum()
perc = people_with_max_hours_over_50K/people_with_max_hours*100

print("Максимальное количество часов в неделю:", max_hours)
print("Количество людей, работающих больше всех:", people_with_max_hours)
print("Процент зарабатывающих более 50К среди трудоголиков:", perc)
```

**10. Посчитайте среднее время работы (*hours-per-week*) зарабатывающих мало и много (*salary*) для каждой страны (*native-country*).**

```python
df = data
per_country_over_50K = (df[df['salary'] == '>50K'].groupby('native-country')['hours-per-week'].mean()
)

per_country_under_50K = (df[df['salary'] == '<=50K'].groupby('native-country')['hours-per-week'].mean()
)

result = pd.concat(
    [per_country_over_50K, per_country_under_50K],
    axis=1,
    keys=['>50K', '<=50K']
)

print(result)
```

**11.Сгруппируйте людей по возрастным группам *young*, *adult*, *retiree*, где:**
* *young* соответствует 16-35 лет
* *adult* - 35-70 лет
* *retiree* - 70-100 лет

**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

```python
df = data

def get_age_group(age):
    if 16 <= age <= 35:
        return 'young'
    elif 35 < age <= 70:
        return 'adult'
    else:
        return 'retiree'

df['AgeGroup'] = df['age'].apply(get_age_group)

df.head()
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

```python
df = data

count_over_50K_by_agegroup = (df[df['salary'] == '>50K'].groupby('AgeGroup')['salary'].count())
top_age_group = count_over_50K_by_agegroup.idxmax()

print(count_over_50K_by_agegroup)
print()
print("Возрастная группа, где чаще всего зарабатывают более 50K:", top_age_group)
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
df = data

occupation_counts = df.groupby('occupation').size()
print(occupation_counts)

def filter_func(group, age=40, hours=5):
    return (group['age'].mean() <= age) and (group['hours-per-week'].min() > hours)

filtered_df = df.groupby('occupation').filter(filter_func)

filtered_counts = filtered_df.groupby('occupation').size()

print()
print("Средний возраст <= 40 и все работают > 5 часов в неделю:")
print(filtered_counts)


```
